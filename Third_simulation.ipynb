```julia
function Wd(path,inital_state, coeffs):
    I=0
   
    for (i,j) in pairs(path)
        n_up=VectorMajoranaSum(MajoranaSum(N_spinful_sites, :nup, i)) #n_1 doublon
        n_dn=VectorMajoranaSum(MajoranaSum(N_spinful_sites, :ndn, i)) #n_1 doublon
        a=overlapwithfock(n_dn, initial_state, coeffs)
        b=overlapwithfock(n_up, initial_state, coeffs)
        I+=(1+a*b)
    end
```
